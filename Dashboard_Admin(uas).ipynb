```python
# =========================
# MENU ADMIN
# =========================
def lihat_semua_booking():
    print("\n=== SELURUH BOOKING ===")
    with koneksi() as conn:
        cur = conn.cursor()
        cur.execute("""
            SELECT b.kode_booking, u.nama, f.nama_fasilitas, b.tanggal,
                   b.jam_mulai, b.jam_selesai, b.status_booking,
                   b.status_bayar, b.total_harga
            FROM booking b
            JOIN users u ON b.id_user = u.id
            JOIN fasilitas f ON b.id_fasilitas = f.id
            ORDER BY b.id DESC
        """)
        data = cur.fetchall()

    if not data:
        print("Belum ada booking.")
        return

    for row in data:
        print("-" * 60)
        print(f"{row[0]} | {row[1]} | {row[2]}")
        print(f"{row[3]} {row[4]}-{row[5]}")
        print(f"Booking: {row[6]} | Bayar: {row[7]} | Rp{row[8]:,}")


def batalkan_booking():
    lihat_semua_booking()
    kode = input("\nKode booking yang dibatalkan: ").strip().upper()

    with koneksi() as conn:
        cur = conn.cursor()
        cur.execute("""
            UPDATE booking
            SET status_booking = 'DIBATALKAN'
            WHERE kode_booking = ?
              AND status_booking NOT IN ('DIBATALKAN', 'SELESAI')
        """, (kode,))
        conn.commit()
        if cur.rowcount:
            print("Booking berhasil dibatalkan.")
        else:
            print("Booking tidak ditemukan atau tidak dapat dibatalkan.")


def menu_admin(user):
    while True:
        print("\n========== DASHBOARD ADMIN ==========")
        print("1. Lihat fasilitas")
        print("2. Tambah fasilitas")
        print("3. Lihat seluruh booking")
        print("4. Validasi pembayaran")
        print("5. Batalkan booking")
        print("6. Laporan penggunaan")
        print("7. Logout")
        pilihan = input("Pilih menu: ").strip()

        if pilihan == "1":
            lihat_fasilitas()
        elif pilihan == "2":
            tambah_fasilitas()
        elif pilihan == "3":
            lihat_semua_booking()
        elif pilihan == "4":
            validasi_pembayaran()
        elif pilihan == "5":
            batalkan_booking()
        elif pilihan == "6":
            laporan_penggunaan()
        elif pilihan == "7":
            print("Admin berhasil logout.")
            break
        else:
            print("Pilihan tidak valid.")
```
